# 🔢 Variablen, Rechnungen und Funktionen

In diesem Notebook lernen wir die Grundbausteine: Werte in Variablen speichern, damit rechnen, und eigene Funktionen schreiben, um Code wiederverwendbar zu machen.

## 1. Variablen und Datentypen

Eine Variable ist ein Name für einen gespeicherten Wert. In Python müssen wir den Typ nicht angeben – er ergibt sich aus dem zugewiesenen Wert.

```python
alter = 25            # int (ganze Zahl)
koerpergroesse = 1.78  # float (Kommazahl)
name = "Alex"          # str (Text)
ist_student = True     # bool (Wahrheitswert: True/False)

print(name, alter, koerpergroesse, ist_student)
print(type(alter), type(koerpergroesse), type(name), type(ist_student))
```

## 2. Einfache Rechnungen

Python kennt die üblichen Rechenoperatoren – inklusive ein paar Besonderheiten (`//` und `%`), die es in vielen anderen Sprachen nicht ganz so übersichtlich gibt.

```python
a = 17
b = 5

print("Addition:      ", a + b)
print("Subtraktion:   ", a - b)
print("Multiplikation:", a * b)
print("Division:      ", a / b)   # Ergebnis ist immer ein float
print("Ganzzahldiv.:  ", a // b)  # rundet nach unten ab (Ergebnis: int)
print("Rest (Modulo): ", a % b)   # was bei der Ganzzahldivision übrig bleibt
print("Potenz:        ", a ** 2)  # a hoch 2
```

### Kurzer Exkurs: Bedingungen

Mit `if` / `elif` / `else` lassen wir Code nur unter bestimmten Bedingungen laufen. Das brauchen wir gleich für unsere erste Funktion.

```python
zahl = 7

if zahl % 2 == 0:
    print(zahl, "ist gerade")
else:
    print(zahl, "ist ungerade")
```

## 3. Funktionen

Eine Funktion fasst wiederverwendbaren Code unter einem Namen zusammen. Mit `def` definieren wir sie, mit `return` geben wir ein Ergebnis zurück.

```python
def ist_gerade(zahl):
    return zahl % 2 == 0


print(ist_gerade(4))
print(ist_gerade(7))
```

Funktionen können mehrere Parameter haben – auch mit **Default-Werten**, die genutzt werden, wenn man beim Aufruf nichts angibt.

```python
def flaeche_rechteck(breite, hoehe=1):
    """Berechnet die Fläche eines Rechtecks."""
    return breite * hoehe


print(flaeche_rechteck(4, 3))   # 12
print(flaeche_rechteck(5))      # hoehe nutzt den Default-Wert 1 -> 5
```

### Ein etwas grösseres Beispiel: Celsius in Fahrenheit umrechnen

```python
def celsius_zu_fahrenheit(grad_celsius):
    return grad_celsius * 9 / 5 + 32


for celsius in [0, 20, 37, 100]:
    print(f"{celsius}\u00b0C = {celsius_zu_fahrenheit(celsius):.1f}\u00b0F")
```
