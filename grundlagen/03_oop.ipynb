# 🧱 Einführung in die objektorientierte Programmierung (OOP)

Bisher haben wir Daten (Variablen) und Verhalten (Funktionen) getrennt behandelt. Bei der **objektorientierten Programmierung** bündeln wir beides in einem **Objekt**:

- Eine **Klasse** ist ein Bauplan (z. B. "Rechteck").
- Ein **Objekt** (auch: **Instanz**) ist ein konkretes Ding, das nach diesem Bauplan gebaut wurde (z. B. "dieses eine Rechteck mit Breite 4 und Höhe 3").
- **Attribute** sind die Daten eines Objekts (z. B. `breite`, `hoehe`).
- **Methoden** sind Funktionen, die zu einem Objekt gehören und mit dessen Daten arbeiten (z. B. `flaeche()`).

## 1. Von der Funktion zur Klasse

Im letzten Notebook hatten wir eine freistehende Funktion `flaeche_rechteck(breite, hoehe)`. Jetzt bauen wir daraus eine Klasse `Rechteck`, die Breite und Höhe selbst "kennt".

```python
class Rechteck:
    def __init__(self, breite, hoehe):
        self.breite = breite
        self.hoehe = hoehe

    def flaeche(self):
        return self.breite * self.hoehe

    def umfang(self):
        return 2 * (self.breite + self.hoehe)
```

**Was passiert hier?**

- `__init__` ist die **Konstruktor-Methode**: Sie wird automatisch aufgerufen, wenn wir ein neues Objekt erzeugen, und legt die Startwerte fest.
- `self` bezeichnet "dieses eine Objekt" – darüber greifen Methoden auf die eigenen Attribute zu. Python übergibt `self` automatisch; wir schreiben es nur in der Definition mit.
- `self.breite = breite` speichert den übergebenen Wert als **Attribut** direkt am Objekt.

```python
r1 = Rechteck(4, 3)
r2 = Rechteck(10, 2)

print("r1 Fläche:", r1.flaeche(), "| Umfang:", r1.umfang())
print("r2 Fläche:", r2.flaeche(), "| Umfang:", r2.umfang())

# jedes Objekt hat seine eigenen Attribute:
print(r1.breite, r2.breite)
```

`r1` und `r2` sind zwei unabhängige Objekte derselben Klasse – jedes mit eigenen Attributwerten, aber demselben "Bauplan" (also denselben Methoden).

## 2. Zustand, der sich verändert: ein Konto

Objekte können auch ihren **Zustand** über die Zeit verändern. Ein klassisches Beispiel ist ein Bankkonto.

```python
class Konto:
    def __init__(self, inhaber, kontostand=0):
        self.inhaber = inhaber
        self.kontostand = kontostand

    def einzahlen(self, betrag):
        self.kontostand += betrag

    def abheben(self, betrag):
        if betrag > self.kontostand:
            print("Nicht genug Guthaben!")
            return
        self.kontostand -= betrag

    def __str__(self):
        return f"Konto von {self.inhaber}: {self.kontostand} CHF"
```

```python
konto = Konto("Sam", kontostand=100)
print(konto)

konto.einzahlen(50)
print(konto)

konto.abheben(30)
print(konto)

konto.abheben(1000)  # zu viel -> Fehlermeldung, Kontostand bleibt gleich
print(konto)
```

`__str__` ist eine sogenannte **Dunder-Methode** ("double underscore") – sie legt fest, was passiert, wenn man das Objekt mit `print()` ausgibt. Es gibt viele solche eingebauten Methoden, mit denen sich Objekte in Python-Standardverhalten einklinken.

## 3. Ausblick: Vererbung und Polymorphismus

Klassen können von anderen Klassen **erben** und dabei Verhalten übernehmen oder überschreiben. So lässt sich z. B. eine allgemeine Klasse `Konto` zu einer spezielleren Klasse `Sparkonto` erweitern, die zusätzlich Zinsen kennt.

```python
class Sparkonto(Konto):
    def __init__(self, inhaber, kontostand=0, zinssatz=0.02):
        super().__init__(inhaber, kontostand)  # ruft __init__ von Konto auf
        self.zinssatz = zinssatz

    def zinsen_gutschreiben(self):
        self.einzahlen(self.kontostand * self.zinssatz)


sparkonto = Sparkonto("Sam", kontostand=1000, zinssatz=0.03)
sparkonto.zinsen_gutschreiben()
print(sparkonto)
```

`Sparkonto` **erbt** `einzahlen()`, `abheben()` und `__str__()` von `Konto` – wir mussten sie nicht neu schreiben. Das ist **Vererbung**.

Wenn verschiedene Klassen dieselbe Methode (z. B. `__str__()` oder `zeichne()`) unterschiedlich umsetzen, aber gleich aufgerufen werden können, nennt man das **Polymorphismus**.
