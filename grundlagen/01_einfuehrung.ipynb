# 🐍 Einführung in Python

Willkommen zum Python-Kurs! Bevor wir loslegen: Was ist Python eigentlich, wo wird es eingesetzt, und wie unterscheidet es sich von anderen Sprachen, denen ihr vielleicht schon begegnet seid?

## Was ist Python?

Python ist eine **universelle, interpretierte Programmiersprache**, die 1991 von Guido van Rossum veröffentlicht wurde. Ein paar Eigenschaften, die sie ausmachen:

- **Interpretiert**: Der Code wird direkt ausgeführt, ohne dass man ihn vorher (wie bei Java oder C#) kompilieren muss. Das macht Ausprobieren und Testen sehr schnell.
- **Dynamisch typisiert**: Man muss den Typ einer Variable nicht angeben – Python erkennt zur Laufzeit, ob etwas eine Zahl, ein Text usw. ist.
- **Lesbare Syntax**: Codeblöcke werden über **Einrückung** (nicht `{ }`) definiert. Das zwingt zu übersichtlichem, einheitlich formatiertem Code.
- **Riesiges Ökosystem**: Für fast jeden Zweck gibt es fertige Bibliotheken – Datenanalyse (`pandas`, `numpy`), Visualisierung (`matplotlib`), Web (`Django`, `FastAPI`), Machine Learning (`scikit-learn`, `PyTorch`) uvm.

Typische Einsatzgebiete: Data Science & Analyse, Automatisierung/Skripte, Web-Backends, Machine Learning, Prototyping.

## Python im Vergleich zu anderen Sprachen

| Sprache | Typsystem | Ausführung | Typischer Einsatz | Wichtigster Unterschied zu Python |
|---|---|---|---|---|
| **Python** | dynamisch | interpretiert | Data Science, Skripte, Web, ML | – |
| **R** | dynamisch | interpretiert | Statistik, Datenanalyse | R ist auf Statistik/Datenanalyse spezialisiert; Python ist universeller einsetzbar (auch Web, Automatisierung, Software allgemein) |
| **C#** | statisch | kompiliert (.NET) | Enterprise-Software, Windows-Apps, Spiele (Unity) | Typen müssen vorab festgelegt werden, Fehler zeigen sich schon beim Kompilieren statt erst beim Ausführen; dafür mehr Schreibaufwand |
| **Java** | statisch | kompiliert (JVM) | Enterprise-Backends, Android | Ähnlich wie C#: strenger, aber sicherer durch Compiler-Prüfung; deutlich mehr "Boilerplate"-Code als Python |
| **SQL** | – | deklarativ | Datenbankabfragen | SQL ist keine Allzwecksprache wie die anderen – man beschreibt *was* man aus einer Datenbank haben will, nicht *wie* der Computer es Schritt für Schritt berechnen soll |

**Merksatz:** Python und R sind beide dynamisch und interpretiert (schnell zu schreiben, langsamer in der Ausführung). C# und Java sind beide statisch und kompiliert (mehr Vorarbeit, dafür mehr Sicherheit und Performance). SQL ist eine ganz eigene Kategorie – deklarativ statt prozedural.

## Vorteile von Python

- ✅ Einfache, lesbare Syntax → schneller Einstieg, auch ohne Informatik-Hintergrund
- ✅ Riesiges Ökosystem an Bibliotheken für praktisch jeden Zweck
- ✅ Interpretiert → sofort testen, kein Kompilierschritt nötig
- ✅ Plattformunabhängig (läuft auf Windows, macOS, Linux)
- ✅ Sehr grosse, aktive Community → viel Dokumentation, Foren, Tutorials

## Nachteile von Python

- ⚠️ Langsamer als kompilierte Sprachen (C#, Java, C++) bei rechenintensiven Aufgaben
- ⚠️ Dynamische Typisierung: Tippfehler bei Typen fallen ohne weitere Massnahmen oft erst zur Laufzeit auf, nicht schon beim Schreiben wie bei Java/C#
- ⚠️ Der "Global Interpreter Lock" (GIL) erschwert echte parallele Ausführung mit mehreren Threads
- ⚠️ Eigenständige Programme für Endanwender ohne installiertes Python (z. B. eine `.exe`) sind aufwändiger zu erstellen als bei kompilierten Sprachen

**Aber:** Diese Nachteile relativieren sich in der Praxis oft:

- **Geschwindigkeit:** Bibliotheken wie `numpy`, `pandas` oder `PyTorch` erledigen die eigentliche Rechenarbeit nicht in Python, sondern in darunterliegendem, hochoptimiertem **C-, C++- oder Fortran-Code**. Python ruft diesen nur auf – man schreibt einfachen Code, bekommt aber (fast) die Geschwindigkeit einer kompilierten Sprache. Zusätzlich lässt sich mit **JIT-Compilern** (z. B. `Numba`, oder alternative Interpreter wie `PyPy`) reiner Python-Code zur Laufzeit in schnellen Maschinencode übersetzen.
- **Dynamische Typisierung:** Optionale **Type Hints** (`def f(x: int) -> str:`) plus Tools wie `mypy` oder `pyright` fangen viele Typfehler schon vor der Ausführung ab – freiwillig nutzbar, aber in professionellen Codebasen mittlerweile Standard.
- **GIL:** Seit Python 3.13 gibt es einen experimentellen "free-threaded" Build ohne GIL (PEP 703), in 3.14 offiziell unterstützt (aber weiterhin nicht Standard). Ausserdem umgeht man den GIL für rechenintensive Aufgaben ohnehin oft über `multiprocessing` oder C-Erweiterungen wie `numpy`, die ihn während der Berechnung freigeben.

Für die meisten Aufgaben im Kursalltag (Datenanalyse, Automatisierung, Skripte) fällt der Geschwindigkeitsunterschied ohnehin kaum ins Gewicht – entscheidender ist meist, wie schnell man selbst zu einer funktionierenden Lösung kommt.
